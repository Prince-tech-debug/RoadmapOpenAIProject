```python
#imports
import os
from dotenv import load_dotenv
from openai import OpenAI
#The import below is for taking the Colab secret keys just comment os and dotenv import and use this one.
# from google.colab import userdata
```

```python
#just comment the one which you dont want to use
load_dotenv(override=True)
OPENAI_API=os.getenv("OPENAI_API_KEY")  #userdata.get('OPENAI_API_KEY')
Gemini_API=os.getenv("GEMINI_API_KEY")  #userdata.get('GEMINI_API_KEY')
if Gemini_API and OPENAI_API:
    print("Both api keys are loaded")
    print(f"GEMINI={Gemini_API[:4]}...")
    print(f"OpenAI={OPENAI_API[:4]}...")
```

```python
client=OpenAI(
    api_key=Gemini_API, 
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)
#For OpenAI API key
Client=OpenAI(api_key=OPENAI_API)

system_prompt="You are an helpfull assistant who helps the users with queries and questions, answer every question honestly in a short paragraph, if you dont know the answer say so."
question="Can you tell what does the Fealing of being unhappy is like."
messages=[{"role":"system","content":system_prompt},{"role":"user","content":question}]
```

### Now this is time to select which model you want to use, just dont run the cells of the model whose API key is not in your environment

```python
#here is the gemini implementation with openAI Lib by using a base url
response=client.chat.completions.create(
    model="gemini-2.5-flash",
    temperature=0.2,    #Try chaning the temp and see how the response changes
    messages=messages
)
```
